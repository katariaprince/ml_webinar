### Indexing and slicing

```python
a = np.array([10,20,30,40,50])

print(a[0])
print(a[-1])
print(a[1:4])
print(a[:3])
print(a[::2])

m = np.array([[1,2,3], [4,5,6]])

print(m[0,2])   #row 0, col 2
print(m[:,1])   #every row col 1 --> 1 feature
print(m[1, :])  #row 1 all col --> one sample
```

### Boolean Masks --> most useful idea in Numpy

```python
scores = np.array([45,82,67,91,38,74])

mask = scores>=60
print(mask)
print(scores[mask])
print(scores[scores>=60].mean())
mid = scores[(scores>=60) & (scores<85)]
print(mid)
```

### Practice Problems

```python
marks = np.array([45, 82, 67, 91, 38, 74, 59, 88])

# 1. How many students scored 60 or above?
print((marks>=60).sum())
# 2. What is the average of only those scores?
print(marks[marks>=60].mean())
# 3. Print the scores that are 60 or above but below 85.
print(scores[(scores>=60) & (scores<85)])
```

```python
# Axis-0 & Axis-1
m=np.arange(12).reshape(3,4)
m.mean(axis=0) #one value per column -> rows vanish
m.sum(axis=1)  #one value per row -> columns vanish
```

```python
m = np.arange(12).reshape(3,4)
# 1. Subtract each column's mean from every row.
# 2. Then divide by each column's standard deviation (m.std(axis=0) ).
# 3. Check your answer: every column should now have a mean of about 0.

z = (m-m.mean(axis=0)) / m.std(axis=0)
print(z)
print(z.mean(axis=0).round(6))
```

#### Reshape & Where

```python
# .reshape(rows, cols)— same numbers, new shape. The total count must match. -1 means "work this oneout for me".
# np.where(condition, a, b)— the array version of if/else.
a = np.array([10,20,30])

np.where(a>15, 1,0)
a.reshape(-1,1)
```
