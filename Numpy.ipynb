```python
# Creating arrays

a = np.array([2,4,6,8,])  #from a list
print(a)

b = np.zeros(5)  #[0. 0. 0. 0. 0.]
print(b)

c = np.ones((2,3))  # 2 rows of 3 cols of 1.
print(c)

d = np.arange(0,10,2)  #[0, 2, 4, 6, 8]
print(d)

e = np.linspace(0,1,5)  #5 evenly spaced 0 to 1
print(e) 
```

### Indexing and slicing

```python
a = np.array([10,20,30,40,50])

print(a[0])
print(a[-1])
print(a[1:4])
print(a[:3])
print(a[::2])

m = np.array([[1,2,3], [4,5,6]])

print(m[0,2])   #row 0, col 2
print(m[:,1])   #every row col 1 --> 1 feature
print(m[1, :])  #row 1 all col --> one sample
```

### Boolean Masks --> most useful idea in Numpy

```python
scores = np.array([45,82,67,91,38,74])

mask = scores>=60
print(mask)
print(scores[mask])
print(scores[scores>=60].mean())
mid = scores[(scores>=60) & (scores<85)]
print(mid)
```

### Practice Problems

```python
marks = np.array([45, 82, 67, 91, 38, 74, 59, 88])

# 1. How many students scored 60 or above?
print((marks>=60).sum())
# 2. What is the average of only those scores?
print(marks[marks>=60].mean())
# 3. Print the scores that are 60 or above but below 85.
print(scores[(scores>=60) & (scores<85)])
```
