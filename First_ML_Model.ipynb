#### Static Program

```python
import pandas as pd

students = pd.read_csv("students.csv")

def will_pass(hours_studied, attendance, prev_marks):
    if hours_studied>=5 and attendance>=75:
        return 1
    return 0

correct = 0
for _, row in students.iterrows():
    my_answer = will_pass(row.hours_studied, row.attendance, row.prev_marks)
    real_answer = row.passed
    if my_answer==real_answer:
        correct+=1

print("Students checked : ", len(students))
print("My rule got right : ", correct)
print("Accuracy : ", round(correct / len(students) * 100, 1), "%")

```

#### ML way - I write no rule

```python
import pandas as pd
from sklearn.linear_model import LogisticRegression

students = pd.read_csv("students.csv")

X = students[["hours_studied", "attendance", "prev_marks"]]  #Features
y = students["passed"]                                       #Target

model = LogisticRegression()
model.fit(X,y)                            #this line is learning

correct = (model.predict(X)==y).sum()

print("students checked : ", len(students))
print("Model got right : ", correct)
print("Accuracy : ", round(correct / len(students) * 100, 1), "%")

# I never wrote a rule. Look what it figured out on its own
for name, weight in zip(X.columns, model.coef_[0]):
    print(f" {name:<15} importance  = {weight:+.3f}")
```

```python
# train test split
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

model = LogisticRegression().fit(X_train, y_train)

print("on data it studied : ", model.score(X_train, y_train))
print("on data it never saw : ", model.score(X_test, y_test))
```

```python
# Ask about one person
one = pd.DataFrame([[8.0, 90, 70]], columns=X.columns)
print(model.predict(one))
print(model.predict_proba(one))        #[[P(fail), P(pass)]]
```
