#### Series

```python
# series -- one column
s = pd.Series([85,72,90,64])
print(s)
```

```python
# you can choose the labels
marks = pd.Series([85,72,90], index=["Aarav", "Priya", "Rohit"])
print(marks["Priya"])
print(marks.mean())
```

#### DataFrame -- The Table

```python
df = pd.DataFrame({
    "name" : ["Aarav", "Priya", "Rohit", "Isha"],
    "branch" : ["CSE", "IT", "CSE", "ECE"],
    "marks" : [85, 72, 90, 64],
})
print(df)
print(df.shape)
print(df.dtypes)
```

##### BUILD YOUR OWN TABLE

```python
# 1. Make a DataFrame of four cities with columns : city, state and population
# 2. Print its shape and  dtypes.
# 3. Print the mean population. 

df = pd.DataFrame({
    "City" : ["Panchkula", "Jaipur", "Roorke", "Rohtak"],
    "State" : ["Haryana", "Rajasthan", "Uttar Pradesh", "Haryana"],
    "Population" : [800000, 1700000, 1100000, 700000]
})
print(df.shape)
print(df.dtypes)
print(df["Population"].mean())
```
