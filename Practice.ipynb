```python
# Pass or fail, without an if
marks = np.array([38, 72, 45, 19, 88, 40, 61, 29, 95, 55])

# 1. Turn it into 1 for pass and 0 for fail, with 40 as the cut-off. Use np.where, not a loop
mask = np.where(marks>=40, 1, 0)
print(mask)
# 2. How many passed?
print(mask.sum())
# 3. What percentage passed?
percentage = (mask.sum() / marks.size) * 100
print(percentage, "%")
```

```python
# Read a score table with axes
scores = np.array([[78, 85, 62],
                   [90, 71, 88],
                   [55, 64, 70],
                   [82, 93, 79]])
# rows = students, columns = subjects

# 1. Each student's total.
student_total = scores.sum(axis=1)
print(student_total)
# 2. Each subject's mean.
subject_mean = scores.mean(axis=0)
print(subject_mean)
# 3. The index of the student with the highest total.
highest_student = np.argmax(student_total)
print(highest_student)
# 4. The index of the subject with the lowest mean.
lowest_mean_subject = np.argmin(subject_mean)
print(lowest_mean_subject)
```

```python
# Clip the outliers
rng = np.random.default_rng(1)
d = rng.normal(50, 10, (8, 3))
d[2, 1] = 150.0      # somebody fat-fingered this
d[6, 0] = -40.0      # and this

# Replace every value that sits more than 2 standard deviations from its own column's mean with that column's
# mean. No loops.
# 1. How many outliers did you find, and where?
# 2. Print the column means before and after.

# col mean and standard deviation
col_mean = d.mean(axis=0)
col_std = d.std(axis=0)

# finding outliers
outliers = abs(d-col_mean) > 2*col_std

# finding outlier position
print("Outlier position (row,column) : ")
print(np.argwhere(outliers))

# replace outliers with column means
cleaned = np.where(outliers, col_mean, d)

# print mean before and after 
print("Means before : ", col_mean)
print("Means after : ", cleaned.mean(axis=0))
```

```python
# Fill the holes
a = np.array([[ 1.,  2., np.nan],
              [ 4., np.nan,  6.],
              [np.nan,  8.,  9.],
              [10., 11., 12.]])

# 1. Compute each column's mean, ignoring the missing values.
col_mean = np.nanmean(a, axis=0)
print(col_mean)
# 2. Replace every missing value with its own column's mean. No loops.
a = np.where(np.isnan(a), col_mean, a)
print(a)
# 3. Prove there are no missing values left.   
print(np.isnan(a).sum())           
```
